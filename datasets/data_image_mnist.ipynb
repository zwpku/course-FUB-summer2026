### MNIST dataset




PyTorch provides many built-in datasets: 

https://docs.pytorch.org/vision/0.21/datasets.html


```python
import torch
import torchvision
from torchvision import transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
```

### prepare the MNIST dataset

Download the data, if it can not be found locally.

```python
# normalize the dataset
transform = transforms.Compose([
    transforms.ToTensor(),
])

# load the MNIST dataset, without normalization
train_dataset = torchvision.datasets.MNIST(root='./data', train=True, transform=transform, download=True)
```

### get the image data and normalize

each image is represented as a 28x28 matrix

```python
# labels
print ("\nLabels:", train_dataset.classes)

# digits of the images
print ("\nClasses:", train_dataset.targets)

# normalize the image data

X = train_dataset.data.float().numpy()
X_mean = X.mean()
X_std = X.std()
X = (X-X_mean) / X_std

print ("\nMean and std:", X_mean, X_std)

print ('\nShape of the data:',  X.shape)

print ('\nMin and Max: ', X.min(), X.max())
```

### display some images from the dataset

```python
num_examples = 10
plt.figure(figsize=(8, 4))
# visualize the images .data
for i in range(num_examples):
    plt.subplot(1, num_examples, i + 1)
    plt.imshow(X[i], cmap='gray')
    plt.axis('off')
    
```

### visualize images according to labels

```python
# number of images to show 
num_examples = 5

# loop over labels
for label in range(10):
    
    # select indices with the matching label
    indices = torch.where(train_dataset.targets == label)[0]
    
    plt.figure(figsize=(4, 2))
    for i in range(num_examples):
        plt.subplot(1, num_examples, i + 1)
        plt.imshow(X[indices[i]], cmap='gray')
        plt.axis('off')
```
