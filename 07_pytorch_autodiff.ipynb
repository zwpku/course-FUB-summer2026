## Simple tests on PyTorch's auto-differentiation functionality

Notice:

1. set requires_grad=True if we want to compute gradient. 
2. call backward() function
3. gradient gets **aggregrated** in the grad field of a tensor.

```python
# need to specify requires_grad=True, in order to calculate derivative 
# with respect to a.
a = torch.tensor([1.0], requires_grad=True)

# define s as a function of a
s = 0.5 * a ** 2

# differentiate
s.backward()

# the derivate is stored in a.grad
print ('a=', a, '\ns=', s, '\ngrad=', a.grad)

s = 0.5 * a ** 2
s.backward()

# the derivate in a.grad gets aggregated.
# to get correct result, we need:
#  a.grad=None 
print ('\n\na=', a, '\ns=', s, '\ngrad=', a.grad)

a.grad = None
s = torch.sin(a)
print ('\n\ns=', s)
s.backward()
print ('a=', a, '\ns=', s, '\ngrad=', a.grad, torch.cos(a))
```

We need to call functions provided by PyTorch. The following code doesn't work.

```python
s = math.sin(a)
s.backward()
```

### Regression using 3rd-order polynomial 

The following PyTorch code learns the function $f(x)=sin(x)$ using a 3rd-order polynomial.

```python
import math

dtype = torch.float

# Create Tensors to hold input and outputs.
# By default, requires_grad=False, which indicates that we do not need to
# compute gradients with respect to these Tensors during the backward pass.
x = torch.linspace(-math.pi, math.pi, 2000, dtype=dtype)
y = torch.sin(x)

# We use a 3rd-order polynomial: 
#     y = a + b x + c x^2 + d x^3
# to approximate sin(x). 

# There are 4 weights: a,b,c,d. 
# Create random Tensors and set requires_grad=True to indicate that we want to compute gradients with
# respect to these Tensors during the learning. 

a = torch.randn((), dtype=dtype, requires_grad=True)
b = torch.randn((), dtype=dtype, requires_grad=True)
c = torch.randn((), dtype=dtype, requires_grad=True)
d = torch.randn((), dtype=dtype, requires_grad=True)

# set the learning rate, or the step-size.
learning_rate = 1e-6

for t in range(2000):    # 2000 iterations in total.
    
    # Forward pass: compute predicted y using operations on Tensors.
    y_pred = a + b * x + c * x ** 2 + d * x ** 3

    # Compute loss using operations on Tensors.
    # Now loss is a Tensor of shape (1,)
    loss = (y_pred - y).pow(2).sum()
    
    # print loss
    if t % 100 == 99:
        # loss.item() gets the scalar value held in the loss.
        print(f'step {t}, loss={loss.item()}')    

    # Use autograd to compute the backward pass. 
    # This call will compute the gradient of loss with respect to all Tensors with requires_grad=True.
    # After this call, a.grad, b.grad. c.grad and d.grad will be Tensors holding
    # the gradient of the loss with respect to a, b, c, d respectively.
    loss.backward()

    # Manually update weights using gradient descent. Wrap in torch.no_grad()
    # because weights have requires_grad=True, but we don't need to track this
    with torch.no_grad():
        a -= learning_rate * a.grad
        b -= learning_rate * b.grad
        c -= learning_rate * c.grad
        d -= learning_rate * d.grad

        # Manually zero the gradients after updating weights (to avoid aggregation)
        a.grad = None
        b.grad = None
        c.grad = None
        d.grad = None

print(f'\nResult: y = {a.item()} + {b.item()} x + {c.item()} x^2 + {d.item()} x^3')

plt.plot(x.detach().numpy(), y.detach().numpy(), label='sin(x)')
plt.plot(x.detach().numpy(), y_pred.detach().numpy(), label='learned polynomial')
plt.legend()
```

### Solve the same regression task, but with torch.nn.Module

```python
# We define the polynomial as a derived class of nn.Module.
class Polynomial3(torch.nn.Module):
    def __init__(self):
        """
        In the constructor we instantiate four parameters and assign them as
        member parameters.
        """
        super().__init__()
        self.a = torch.nn.Parameter(torch.randn(()))
        self.b = torch.nn.Parameter(torch.randn(()))
        self.c = torch.nn.Parameter(torch.randn(()))
        self.d = torch.nn.Parameter(torch.randn(()))
        
    # We always need to define a forward function, telling how the model maps the input to output. 
    def forward(self, x):
        """
        In the forward function we accept a Tensor of input data and we must return
        a Tensor of output data. 
        
        We can use Modules defined in the constructor as well as arbitrary operators on Tensors.
        """
        return self.a + self.b * x + self.c * x ** 2 + self.d * x ** 3

    def string(self):
        """
        Just like any class in Python, we can also define custom method on PyTorch modules
        """
        return f'y = {self.a.item()} + {self.b.item()} x + {self.c.item()} x^2 + {self.d.item()} x^3'


# Create Tensors to hold input and outputs.
# This is our data (with labels)
x = torch.linspace(-math.pi, math.pi, 2000)
y = torch.sin(x)

# Construct our model by instantiating the class defined above
model = Polynomial3()

# Construct our loss function. 
criterion = torch.nn.MSELoss(reduction='sum')

# Define the optimizer. The call to model.parameters() in the SGD constructor returns the learnable parameters 
# (defined with torch.nn.Parameter) which are members of the model.
optimizer = torch.optim.SGD(model.parameters(), lr=1e-6) 

for t in range(2000):
    
    # Forward pass: Compute predicted y by passing x to the model
    # This will call the forward() function defined above.
    y_pred = model(x)

    # Compute and print loss
    loss = criterion(y_pred, y)
    
    if t % 100 == 99:
        print(f'step {t}, loss={loss.item()}')    

    # Zero gradients, perform a backward pass, and update the weights.
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print(f'Result: {model.string()}')

plt.plot(x.detach().numpy(), y.detach().numpy())
plt.plot(x.detach().numpy(), y_pred.detach().numpy())

plt.plot(x.detach().numpy(), y.detach().numpy(), label='sin(x)')
plt.plot(x.detach().numpy(), y_pred.detach().numpy(), label='learned polynomial')
plt.legend()
```
