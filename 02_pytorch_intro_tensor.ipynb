### Tensor initialization

we can define PyTorch tensors and examine their types.

```python
x = torch.zeros(3, 4)
print(type(x))
print(x)
print (x.shape)

x = torch.ones(3, 4)
print(x)

torch.manual_seed(1721)
random = torch.rand(2, 3)

print(random)
```

### Basic math operation

Different operations can be applied to tensors.

```python
ones = torch.zeros(2, 2) + 1
twos = torch.ones(2, 2) * 2
threes = (torch.ones(2, 2) * 7 - 1) / 2
fours = twos ** 2    # element-wise
sqrt2s = twos ** 0.5

print(ones)
print(twos)
print(threes)
print(fours)
print(sqrt2s)
```

### Change shape 

Tensor shapes can be changed. 

```python
x = torch.ones(3, 4)
print(x.shape)
y = x.reshape(2,6)
print (y.shape)
z = x.unsqueeze(2)
print (z.shape)
w = z.squeeze()
print (w.shape)
```

### Broadcasting

Broadcasting rules specify how to perform operations on tensors of different types. 


```python
a =     torch.ones(4, 3, 2)

b = a * torch.rand(   3, 2) # 3rd & 2nd dims identical to a, dim 1 absent
print(b.shape)

c = a * torch.rand(   3, 1) # 3rd dim = 1, 2nd dim identical to a
print(c.shape)

d = a * torch.rand(   1, 2) # 3rd dim identical to a, 2nd dim = 1
print(d.shape)
```

### Linear Algebra 

Various linear algebra operations are provided.

```python

v1 = torch.tensor([1., 0., 0.])         # x unit vector
v2 = torch.tensor([0., 1., 0.])         # y unit vector
m1 = torch.rand(2, 2)                   # random matrix
m2 = torch.tensor([[3., 0.], [0., 3.]]) # three times identity matrix

print(torch.linalg.cross(v2, v1)) # negative of z unit vector (v1 x v2 == -v2 x v1)
print('m1:', m1)

m3 = torch.linalg.matmul(m1, m2)
print('m3:', m3)                  # 3 times m1

U, S, Vh = torch.linalg.svd(m3)
print ("U, S, Vh=", U, S, Vh)
```

### PyTorch tensors can be converted to/from numpy arrays.

```python
numpy_array = np.ones((2, 3))
print(numpy_array)

# from numpy to pytorch tensor
pytorch_tensor = torch.from_numpy(numpy_array)
print(pytorch_tensor)

# from pytorch tensor to numpy
numpy_array_1 = pytorch_tensor.numpy()
print(numpy_array_1)
```
